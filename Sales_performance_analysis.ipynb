```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
import pandas as pd

sales = pd.read_csv("sales.csv", encoding='latin1')
sales.head()
```

```python
#DATA CLEANNG
sales.isnull().sum()
```

```python
print(list(sales.columns))
```

```python
sales['Order_Date'] = pd.to_datetime(
    sales['Order_Date'],
    format='%d/%m/%Y'
)
```

```python
total_revenue = sales['Sales'].sum()
print("Total Revenue:", total_revenue)
```

```python
total_orders = sales['Order_ID'].nunique()
print("Total Orders:", total_orders)
```

```python
avg_order_value = sales['Sales'].sum() / sales['Order_ID'].nunique()

print("Average Order Value:", round(avg_order_value,2))
```

```python
region_sales = sales.groupby('Region')['Sales'].sum().sort_values(ascending=False)

print(region_sales)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
region_sales.plot(kind='bar')
plt.title("Sales by Region")
plt.ylabel("Revenue")
plt.show()
```

```python
category_sales = sales.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))
category_sales.plot(kind='bar')
plt.title("Sales by Category")
plt.ylabel("Revenue")
plt.show()
```

```python
subcategory_sales = sales.groupby('Sub_Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
subcategory_sales.plot(kind='bar')
plt.title("Sales by Sub-Category")
plt.ylabel("Revenue")
plt.show()
```

```python
top_products = sales.groupby('Product_Name')['Sales'].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
plt.figure(figsize=(12,6))
top_products.plot(kind='bar')
plt.title("Top 10 Products")
plt.ylabel("Revenue")
plt.show()
```

```python
bottom_products = sales.groupby('Product_Name')['Sales'].sum().sort_values().head(10)

print(bottom_products)
```

```python
plt.figure(figsize=(12,6))
bottom_products.plot(kind='bar')
plt.title("Bottom 10 Products")
plt.ylabel("Revenue")
plt.show()
```

```python
sales['Order_Date'] = pd.to_datetime(
    sales['Order_Date'],
    format='%d/%m/%Y'
)

sales['Year'] = sales['Order_Date'].dt.year
sales['Month'] = sales['Order_Date'].dt.month
sales['Month_Name'] = sales['Order_Date'].dt.month_name()
```

```python
print(sales[['Order_Date','Year','Month','Month_Name']].head())
```

```python
monthly_sales = sales.groupby('Month_Name')['Sales'].sum()

print(monthly_sales)
```

```python
import matplotlib.pyplot as plt

monthly_sales = sales.groupby('Month')['Sales'].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
yearly_sales = sales.groupby('Year')['Sales'].sum()

plt.figure(figsize=(8,5))
yearly_sales.plot(marker='o')
plt.title("Yearly Sales Trend")
plt.ylabel("Revenue")
plt.show()
```

```python
top_states = sales.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
top_states.plot(kind='bar')
plt.title("Top 10 States by Revenue")
plt.ylabel("Revenue")
plt.show()
```

```python
segment_sales = sales.groupby('Segment')['Sales'].sum()

plt.figure(figsize=(8,5))
segment_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales by Customer Segment")
plt.ylabel('')
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(6,4))

sns.heatmap(
    sales[['Sales']].corr(),
    annot=True
)

plt.show()
```
